# Rule Manager Interpreter — end-to-end demo

This notebook builds the synthetic banking database, displays and runs the original SQL rule, compiles that rule to the versioned JSON DSL, and executes the DSL with both DuckDB and real PySpark. The final check proves that all three result sets agree.

```python
from pathlib import Path
import html
import json
import subprocess
import sys

import duckdb
from IPython.display import Code, HTML, Markdown, display


def find_project_root() -> Path:
    candidates = [Path.cwd(), *Path.cwd().parents]
    for candidate in candidates:
        if (candidate / 'pyproject.toml').exists() and (candidate / 'demo').exists():
            return candidate
    raise RuntimeError('Run this notebook from inside the rule-manager-interpreter repository.')


ROOT = find_project_root()
DEMO = ROOT / 'demo'
CATALOG = DEMO / 'catalog.json'
SQL_RULE = DEMO / 'rules' / 'credit_decisions.sql'
DSL_FILE = DEMO / 'generated' / 'credit_decisions.dsl.json'


def run_cli(*arguments: str) -> dict:
    command = [sys.executable, '-m', 'rule_interpreter.cli', *map(str, arguments)]
    completed = subprocess.run(command, cwd=ROOT, text=True, capture_output=True, check=False)
    if completed.returncode != 0:
        raise RuntimeError(f"Command failed: {' '.join(command)}\n{completed.stderr}")
    return json.loads(completed.stdout) if completed.stdout.strip() else {}


def show_result(result: dict, title: str) -> None:
    columns = result['columns']
    header = ''.join(f'<th>{html.escape(str(column))}</th>' for column in columns)
    rows = []
    for row in result['rows']:
        cells = ''.join(f'<td><code>{html.escape(json.dumps(value, ensure_ascii=False))}</code></td>' for value in row)
        rows.append(f'<tr>{cells}</tr>')
    display(HTML(f'<h3>{html.escape(title)}</h3><p>{result["row_count"]} rows</p><table><thead><tr>{header}</tr></thead><tbody>{"".join(rows)}</tbody></table>'))

print(f'Project root: {ROOT}')
```

## 1. Build the sample DuckDB database and shared Parquet fixtures

```python
build = run_cli('build-demo', '--root', str(DEMO))
display(build)
```

## 2. Display the original script

This is equivalent to inspecting the file with `cat demo/rules/credit_decisions.sql`, but rendered with SQL highlighting.

```python
source_sql = SQL_RULE.read_text(encoding='utf-8')
display(Code(source_sql, language='sql'))
```

## 3. Execute the original script directly

Direct source execution is the trusted demo oracle. Production rule execution consumes validated DSL instead.

```python
with duckdb.connect(str(DEMO / 'banking.duckdb'), read_only=True) as connection:
    cursor = connection.execute(source_sql)
    direct_result = {
        'engine': 'direct-duckdb',
        'columns': [column[0] for column in cursor.description],
        'rows': [list(row) for row in cursor.fetchall()],
    }
direct_result['row_count'] = len(direct_result['rows'])
show_result(direct_result, 'Original SQL result')
```

## 4. Compile SQL to DSL with the CLI

```python
DSL_FILE.parent.mkdir(parents=True, exist_ok=True)
run_cli(
    'sql-to-dsl', str(SQL_RULE),
    '--catalog', str(CATALOG),
    '--output', str(DSL_FILE),
    '--rule-id', 'credit-decisions',
)
dsl = json.loads(DSL_FILE.read_text(encoding='utf-8'))
display({
    'dsl_file': str(DSL_FILE),
    'dsl_version': dsl['dsl_version'],
    'rule': dsl['rule'],
    'inputs': dsl['inputs'],
    'integrity': dsl['integrity'],
})
```

## 5. Display the generated JSON DSL

```python
display(Code(json.dumps(dsl, indent=2, sort_keys=True), language='json'))
```

## 6. Execute the DSL with DuckDB

```python
duckdb_dsl_result = run_cli(
    'execute', str(DSL_FILE),
    '--engine', 'duckdb',
    '--catalog', str(CATALOG),
)
show_result(duckdb_dsl_result, 'DSL result — DuckDB executor')
```

## 7. Execute the same DSL with real PySpark

The Spark executor loads the catalog-approved Parquet fixtures and never receives the original SQL source.

```python
spark_dsl_result = run_cli(
    'execute', str(DSL_FILE),
    '--engine', 'spark',
    '--catalog', str(CATALOG),
)
show_result(spark_dsl_result, 'DSL result — PySpark executor')
```

## 8. Compare the direct script and DSL outputs

```python
def semantic_result(result: dict) -> dict:
    return {key: result[key] for key in ('columns', 'rows', 'row_count')}

comparison = {
    'direct_equals_duckdb_dsl': semantic_result(direct_result) == semantic_result(duckdb_dsl_result),
    'direct_equals_spark_dsl': semantic_result(direct_result) == semantic_result(spark_dsl_result),
    'duckdb_dsl_equals_spark_dsl': semantic_result(duckdb_dsl_result) == semantic_result(spark_dsl_result),
}
assert all(comparison.values()), comparison
display(comparison)
display(Markdown('### ✅ The original script and both DSL executors produced identical results.'))
```
